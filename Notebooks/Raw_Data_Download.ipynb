```python
ham_path = kagglehub.dataset_download("kmader/skin-cancer-mnist-ham10000")
pad_path = kagglehub.dataset_download("mahdavi1202/skin-cancer")
bcn_path = kagglehub.dataset_download("pasutchien/bcn20000")

# 2. The Rescue Protocol
def rescue_metadata(source_dir, final_dir):
    os.makedirs(final_dir, exist_ok=True)
    csv_count = 0
    
    for root, dirs, files in os.walk(source_dir):
        for file in files:
            if file.lower().endswith('.csv'):
                orig_path = os.path.join(root, file)
                final_path = os.path.join(final_dir, file)
                
                # Copy the CSV from the hidden cache to your project folder
                shutil.copy2(orig_path, final_path)
                print(f"Rescued: {file} --> {final_dir}")
                csv_count += 1
                
    if csv_count == 0:
        print(f"WARNING: No CSV found in {source_dir}")

# 3. Execute the rescue
print("\nExecuting Metadata Rescue...")
rescue_metadata(ham_path, "../data/raw/HAM10000")
rescue_metadata(pad_path, "../data/raw/PAD_UFES_20")
rescue_metadata(bcn_path, "../data/raw/BCN20000")
print("\nRescue Complete. Your metadata is now in place.")
```
